```python
import numpy as np



result = np.random.choice([1, 2, 3, 4, 5, 6], size=10)
print(result)
```

```python

result = np.random.choice([1, 2, 3, 4, 5, 6], size=1000)


sixes = np.sum(result == 6)

print("Number of 6s:", sixes)
print("Probability of rolling a 6:", sixes / 1000)
```

```python
import matplotlib.pyplot as plt


counts = [np.sum(result == i) for i in range(1, 7)]


plt.bar([1, 2, 3, 4, 5, 6], counts)
plt.xlabel("Die Face")
plt.ylabel("Count")
plt.title("Dice Roll Simulation")
plt.show()
```
