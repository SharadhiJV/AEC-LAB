```python
import numpy as np

marks = np.array([65, 72, 33, 44])
print("Mean:", marks.mean())
print("Maximum:", marks.max())
```

```python
import pandas as pd

data={
    "Name":["A","B","C","D","E"],
    "Marks":[65,72,81,55,90]

}
df=pd.DataFrame(data)
print(df)
print(df["Marks"].mean())
```

```python
import matplotlib.pyplot as plt

plt.bar(df["Name"],df["Marks"])
plt.xlabel("student")
plt.ylabel("Mraks")
plt.title("Student Marks")
plt.show()
```

```python
result=np.random.choice(['Head"',"Tail"],size=10)
print(result)
```

```python
result=np.random.choice(["Head","Tail"],size=1000)
head=np.sum(result=="Head")
print("Heads:",head)
print("probablity of head:",head/1000)
```

```python
counts={
    np.sum(result=="Head"),
    np.sum(result=="Tail")

}
plt.bar(["Head","Tail"],counts)
plt.xlabel("Outcome")
plt.ylabel("count")
plt.title("coin Toss Simulation")
plt.show()
```
